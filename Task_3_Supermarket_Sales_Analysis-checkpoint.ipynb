```python
branch_sales = df.groupby('Branch')['Total'].sum()

plt.figure(figsize=(8,5))
branch_sales.plot(kind='bar')

plt.title('Total Sales by Branch')
plt.xlabel('Branch')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
product_sales = df.groupby('Product line')['Total'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))
product_sales.plot(kind='bar')

plt.title('Total Sales by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
customer_sales = df.groupby('Customer type')['Total'].sum()

plt.figure(figsize=(8,5))
customer_sales.plot(kind='bar')

plt.title('Total Sales by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
gender_sales = df.groupby('Gender')['Total'].sum()

plt.figure(figsize=(8,5))
gender_sales.plot(kind='bar')

plt.title('Total Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
payment_sales = df.groupby('Payment')['Total'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
payment_sales.plot(kind='bar')

plt.title('Total Sales by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
df['Month'] = df['Date'].dt.to_period('M')
```

```python
monthly_sales = df.groupby('Month')['Total'].sum()

monthly_sales
```

```python
plt.figure(figsize=(10,5))
monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
product_quantity = df.groupby('Product line')['Quantity'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))
product_quantity.plot(kind='bar')

plt.title('Total Quantity Sold by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df['Total'], bins=30)

plt.title('Distribution of Total Sales')
plt.xlabel('Total Sales')
plt.ylabel('Number of Transactions')

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df['Rating'], bins=10)

plt.title('Distribution of Customer Ratings')
plt.xlabel('Customer Rating')
plt.ylabel('Number of Customers')

plt.show()
```

```python
rating_by_product = df.groupby('Product line')['Rating'].mean().sort_values(ascending=False)

rating_by_product
```

```python
plt.figure(figsize=(10,6))

rating_by_product.plot(kind='bar')

plt.title('Average Customer Rating by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Average Rating')
plt.xticks(rotation=45, ha='right')

plt.show()
```

```python
numeric_data = df[['Unit price', 'Quantity', 'Tax 5%', 'Total', 'cogs', 'gross income', 'Rating']]

correlation = numeric_data.corr()

correlation
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(correlation, annot=True)

plt.title('Correlation Between Numerical Variables')

plt.show()
```

## Correlation Analysis - Interpretation

Correlation analysis was performed to understand the relationships between numerical variables.

- Quantity and Total showed a strong positive relationship of 0.71.
- Unit Price and Total showed a positive relationship of 0.63.
- Rating and Total showed almost no relationship.
- Rating and Quantity also showed almost no relationship.
- Tax 5%, Total, COGS, and Gross Income showed very high correlations because these variables are mathematically related.

This analysis helped us understand the relationships between different numerical factors affecting supermarket sales.

## Final Business Insights

The analysis of the supermarket sales dataset revealed the following important findings:

1. **Branch Performance:** Branch C generated the highest total sales of approximately 110,568.71, followed by Branch A and Branch B.

2. **Product Line Performance:** Food and beverages generated the highest sales of approximately 56,144.84, while Health and beauty generated the lowest sales.

3. **Customer Type:** Member customers contributed approximately 164,223.44 in sales, which was higher than normal customers.

4. **Gender Analysis:** Female customers generated higher sales of approximately 167,882.93 compared to male customers.

5. **Payment Method:** Cash was the most used payment method in terms of total sales, followed by E-wallet and Credit card.

6. **Monthly Sales Trend:** January 2019 recorded the highest sales among the three months. Sales decreased in February and increased again in March.

7. **Customer Ratings:** Food and beverages had the highest average customer rating among the product lines, while Home and lifestyle had the lowest.

8. **Correlation Analysis:** Quantity and Total showed a strong positive relationship. Unit Price also showed a positive relationship with Total, while customer ratings had almost no linear relationship with sales.
